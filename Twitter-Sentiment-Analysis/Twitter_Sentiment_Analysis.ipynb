# Twitter Sentiment Analysis using Python

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import BernoulliNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv(r"E:\1Full_Stack_Academy\CSV file\training.1600000.processed.noemoticon.csv\training.1600000.processed.noemoticon.csv",encoding='latin-1',header=None)
df = df[[0,5]]
df.columns = ['polarity', 'text']
print(df.head())
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
print(df.shape)
print(df.head())
```

```python
print(df.head())
print(df.info())
```

```python
df = df[df.polarity != 2]
df['polarity'] = df['polarity'].map({0: 0,4: 1})
print(df['polarity'].value_counts())
```

```python
def clean_text(text):
    return text.lower()
df['clean_text'] = df['text'].apply(clean_text)
print(df[['text','clean_text']].head())
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    df['clean_text'],
    df['polarity'],
    test_size=0.2,
    random_state=42
)
```

```python
print('Train size:', len(X_train))
print('Test size:', len(X_test))
```

```python
print('Train size:', len(Y_train))
print('Test size:', len(Y_test))
```

```python
vectorizer = TfidfVectorizer(max_features=5000, ngram_range= (1,2))

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

```

```python
print("TF-IDF shape (train):", X_train_tfidf.shape)
print("TF-IDF shape (test):", X_test_tfidf.shape)
```

```python
bnb = BernoulliNB()
bnb.fit(X_train_tfidf, Y_train)
bnb_pred = bnb.predict(X_test_tfidf)
```

```python
print(accuracy_score(Y_test, bnb_pred))
```

```python
print(classification_report(Y_test,bnb_pred))
```

```python
svm = LinearSVC(max_iter=1000)
svm.fit(X_train_tfidf, Y_train)
svm_pred = svm.predict(X_test_tfidf)
```

```python
print(accuracy_score(Y_test, svm_pred))
print(classification_report(Y_test,svm_pred))
```

```python
logreg = LogisticRegression(max_iter=100)
logreg.fit(X_train_tfidf,Y_train)

logreg_pred = logreg.predict(X_test_tfidf)
```

```python
print(accuracy_score(Y_test, logreg_pred))
print(classification_report(Y_test,logreg_pred))
```

```python
print("Bernoulli Naive Bayes:",
      accuracy_score(Y_test, bnb_pred))

print("SVM:",
      accuracy_score(Y_test, svm_pred))

print("Logistic Regression:",
      accuracy_score(Y_test, logreg_pred))
```
